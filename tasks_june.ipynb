```python
#Task 9: Monthly Revenue Trend
```

```python
orders_products["order_date"] = pd.to_datetime(
    orders_products["order_date"]
)

orders_products["month"] = (
    orders_products["order_date"]
    .dt.month_name()
)

monthly_revenue = (
    orders_products
    .groupby("month")
    ["revenue"]
    .sum()
    .reset_index()
)

monthly_revenue
```

```python
#Task 17: Build a Star Schema
```

```python
#dim_customer Create
dim_customer = customers.copy()

dim_customer["customer_key"] = range(
    1,
    len(dim_customer)+1
)

dim_customer = dim_customer[
    [
        "customer_key",
        "customer_id",
        "customer_name",
        "city",
        "state",
        "signup_date"
    ]
]

dim_customer
```

```python
#dim_product Create
dim_product = products.copy()

dim_product["product_key"] = range(
    1,
    len(dim_product)+1
)

dim_product = dim_product[
    [
        "product_key",
        "product_id",
        "product_name",
        "category",
        "price"
    ]
]

dim_product
```

```python
#dim_date Create
dim_date = fact_orders.copy()

dim_date["order_date"] = pd.to_datetime(
    dim_date["order_date"]
)

dim_date = dim_date[
    ["order_date"]
].drop_duplicates()

dim_date["date_key"] = range(
    1,
    len(dim_date)+1
)

dim_date["month"] = (
    dim_date["order_date"]
    .dt.month
)

dim_date["year"] = (
    dim_date["order_date"]
    .dt.year
)

dim_date["quarter"] = (
    dim_date["order_date"]
    .dt.quarter
)

dim_date
```

```python
#Revenue Column Create
orders_products["revenue"] = (
    orders_products["quantity"]
    *
    orders_products["price"]
)
```

```python
#Fact Table Create
fact_table = pd.merge(
    orders_products,
    dim_customer[
        ["customer_key","customer_id"]
    ],
    on="customer_id"
)

fact_table = pd.merge(
    fact_table,
    dim_product[
        ["product_key","product_id"]
    ],
    on="product_id"
)

fact_table = pd.merge(
    fact_table,
    dim_date[
        ["date_key","order_date"]
    ],
    on="order_date"
)

fact_table
```

```python
#Required Columns
```

```python
fact_orders_star = fact_table[
    [
        "order_id",
        "customer_key",
        "product_key",
        "date_key",
        "quantity",
        "revenue"
    ]
]

fact_orders_star
```
